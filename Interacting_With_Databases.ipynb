```python
import sqlite3
import pandas as pd
```

```python
con = sqlite3.connect("ECom.sqlite")
```

```python
query="""CREATE TABLE users (
    user_id INTEGER PRIMARY KEY,
    username VARCHAR(50) NOT NULL,
    email VARCHAR(100) UNIQUE,
    age INTEGER,
    created_at TIMESTAMP DEFAULT CURRENT_TIMESTAMP
);"""
```

```python
con.execute(query)
```

```python
 con.commit()
```

```python

insert_query = """
INSERT INTO users (username, email, age)
VALUES
    ('alice_w', 'alice@example.com', 28),
    ('bob_dev', 'bob@example.com', 34),
    ('charlie_k', 'charlie@example.com', 22);
"""

```

```python
con.execute(insert_query)
con.commit()
```

```python
cursor = con.execute("SELECT * FROM users")
```

```python
rows = cursor.fetchall()
```

```python

query = "SELECT * FROM users WHERE age > ?"

# 2. Pass the filter value inside a tuple (25,)
cursor = con.execute(query, (25,))
rows = cursor.fetchall()

# 3. View filtered rows
print(rows)
```

```python
import sqlalchemy as sqla
from sqlalchemy import create_engine, text
```

```python
engine = sqla.create_engine("sqlite:///Ecom1check.sqlite")
```

```python
inspector = sqla.inspect(engine)
table_names = inspector.get_table_names()

print("Available tables:", table_names)
```

```python
engine = create_engine("sqlite:///ECom.sqlite")

with engine.connect() as connection:
    query = text("SELECT * FROM users LIMIT 5;")
    df = pd.read_sql_query(query, connection)

df
```
